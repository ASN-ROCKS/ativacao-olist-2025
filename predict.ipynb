```python
import mlflow

# OBTENDO O MODELO
versions = mlflow.search_model_versions(filter_string="name='analytics.olist.ativacao_t5'")
version = max([int(i.version) for i in versions])
model = mlflow.sklearn.load_model(f"models:///analytics.olist.ativacao_t5/{version}")

# OBTENDO NOVOS DADOS
df = spark.table("analytics.olist.abt_ativacao").toPandas()

proba_churn = model.predict_proba(df[model.feature_names_in_])
proba_churn = proba_churn[:,1]

df_results = df[['idVendedor']].copy()
df_results['prob_churn'] = proba_churn
df_results

# 0	9c068d10aca38e85c50202e17b4a7e88	0.342134
# 1	0ea22c1cfbdc755f86b9b54b39c16043	0.130030
```
